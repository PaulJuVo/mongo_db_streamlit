```python
import json
import pandas as pd
import pathlib
import itertools as it
import pprint as p
```

```python
#file_type: list[str] = ["Profile", "IncomeStatement", "eod_prices"]
file_type: list[str] = ["Profile"]
company_names = ["AAPL"]

base_path = pathlib.Path("/Users/paulvogt/mongo_db_streamlit/tmp")
path =  base_path / "raw/total_sp_data"

cartesian_name = it.product([path.as_posix()], company_names, file_type)
file_paths = list(map(lambda a :  a[0] + "/" + a[1] + "_" + a[2] + ".json", cartesian_name))

print(*file_paths, sep=", ")
```

```python
i = 1
for f_path in file_paths:
    with open(f_path) as file:
        data = json.load(file)
        df = pd.json_normalize(data)
        #hist_df = df["historical"]
        #test = pd.json_normalize(hist_df[0])
        df.to_json(base_path.as_posix() + f"/test/{i}_import.json")
        i += 1
    
```
