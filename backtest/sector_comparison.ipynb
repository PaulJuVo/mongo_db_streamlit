```python
import pandas as pd
from pathlib import Path
import os
import plotly.express as px

os.chdir(Path.cwd().parent)

save_path= "backtest/attachments"
```

```python
df = pd.read_csv("backtest/data/backtest_data.csv")
df["horizon"] = df["sellyear"] - df["buyyear"]
df_horizon = df.drop(axis=1,labels=["sellyear", "buyyear"])
df_grouped = df_horizon.groupby(by=["sector", "strategy", "horizon"], as_index=False).median()
df_grouped
```

```python
df_mom = df_grouped[df_grouped["strategy"] == "momentum_score" ].drop(["strategy"], axis=1)
df_mom_piv = df_mom.pivot(index="sector", columns="horizon")
df_mom_piv
```

```python
df_val = df_grouped[df_grouped["strategy"] == "value_score" ].drop(["strategy"], axis=1)
df_val_piv = df_val.pivot(index="sector", columns="horizon")
df_val_piv
```

```python
df_value = df_grouped[df_grouped["strategy"] == "value_score"].set_index(["sector", "horizon"])["rendite"]
df_mom   = df_grouped[df_grouped["strategy"] == "momentum_score"].set_index(["sector", "horizon"])["rendite"]


df_delta = (df_value - df_mom).reset_index()
df_delta.columns = ["sector", "horizon", "rendite"]

df_delta
```

```python
horizons = [1,3,5]
df_final = df_delta[df_delta["horizon"].isin(horizons)]
```

```python
from pathlib import Path
import plotly.express as px
import plotly.graph_objects as go

BLUE = "#4E79A7"      # positiv = Value
ORANGE = "#F28E2B"    # negativ = Momentum

for h in horizons:

    df_filtered = (
        df_final[df_final["horizon"] == h]
        .sort_values("rendite", ascending=True)
        .copy()
    )

    df_filtered["Farbe"] = df_filtered["rendite"].apply(
        lambda x: BLUE if x >= 0 else ORANGE
    )

    fig = px.bar(
        df_filtered,
        x="rendite",
        y="sector",
        orientation="h",
    )

    fig.update_traces(
        marker_color=df_filtered["Farbe"],
        marker_line_width=0,
        width=0.65,
        text=df_filtered["rendite"].map(lambda x: f"{x:+.2f}"),
        textposition="outside",
        textfont=dict(size=14, family="Latin Modern Roman, Times New Roman, serif"),
        cliponaxis=False,          # <-- verhindert das Abschneiden der Labels beim Export
    )

    fig.add_vline(
        x=0,
        line_width=1.5,
        line_color="#333333"
    )

    # Symmetrische Achse mit mehr Puffer für die Labels
    max_abs = abs(df_filtered["rendite"]).max()

    fig.update_xaxes(
        title="Rendite-Delta (Value − Momentum, in Prozentpunkten)",
        range=[-max_abs * 1.35, max_abs * 1.35],   # mehr Puffer statt 1.1
        zeroline=False,
        gridcolor="#e6e6e6",
        showline=True,
        linecolor="#333333",
        ticks="outside",
    )

    fig.update_yaxes(
        title="",
        categoryorder="array",
        categoryarray=df_filtered["sector"],
        showline=True,
        linecolor="#333333",
        ticks="outside",
        automargin=True,           # verhindert abgeschnittene Sektornamen
    )

    fig.update_layout(
        title=f"Rendite-Delta nach Sektor ({h}-Jahres-Horizont)",
        title_x=0.5,
        title_font=dict(size=17, family="Latin Modern Roman, Times New Roman, serif"),
        font=dict(size=14, family="Latin Modern Roman, Times New Roman, serif", color="#222222"),
        width=950,
        height=580,
        template="simple_white",
        margin=dict(l=40, r=60, t=80, b=50),
        showlegend=True,
        legend=dict(
            orientation="h",
            yanchor="bottom", y=1.02,
            xanchor="center", x=0.5,
            title=None,
        ),
        plot_bgcolor="white",
        paper_bgcolor="white",
    )

    # Dummy-Traces nur für eine saubere Legende (Farbe = Strategie, nicht Vorzeichen)
    fig.add_trace(go.Bar(x=[None], y=[None], marker_color=BLUE, name="Value outperformt"))
    fig.add_trace(go.Bar(x=[None], y=[None], marker_color=ORANGE, name="Momentum outperformt"))

    fig.show()

    output_dir = Path(save_path) / "sector_comparison"
    output_dir.mkdir(parents=True, exist_ok=True)

    file_path = output_dir / f"{h}_year_sorted_bar.png"
    fig.write_image(file_path, scale=3)
```
